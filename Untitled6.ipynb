```python

from ultralytics import YOLO
from IPython.display import display
from PIL import Image
import glob
import os

# 1. Load your trained model
model = YOLO("/content/traffic_project/vehicle_training/weights/best.pt")

# 2. Find images in your validation dataset
image_files = glob.glob("/content/vehicle_dataset/valid/images/*")

# 3. Check whether images are available
print("Images found:", len(image_files))

if len(image_files) > 0:
    # Select the first image
    image_path = image_files[0]
    print("Testing image:", os.path.basename(image_path))

    # 4. Run vehicle detection
    results = model.predict(
        source=image_path,
        conf=0.25,
        save=True
    )

    # 5. Display detection result
    annotated_image = results[0].plot()
    display(Image.fromarray(annotated_image[..., ::-1]))

    # 6. Print detected objects
    for box in results[0].boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        print(
            f"Detected: {model.names[class_id]} | "
            f"Confidence: {confidence:.2%}"
        )

    print("Test completed!")
    print("Annotated results saved in:", results[0].save_dir)

else:
    print("No images found. Check your dataset folder path.")
```

```python

from IPython.display import display
from PIL import Image
import glob
import os

image_files = glob.glob("/content/vehicle_dataset/valid/images/*")

if not image_files:
    print("No images found in the validation folder.")
else:
    for image_path in image_files[:5]:
        results = model.predict(source=image_path, conf=0.25, verbose=False)

        print("\nImage:", os.path.basename(image_path))
        print("Number of detections:", len(results[0].boxes))

        annotated_image = results[0].plot()
        display(Image.fromarray(annotated_image[..., ::-1]))
```
